````python
# Cell 2: System Telemetry & Sandboxed Code Execution Harness

def get_peak_rss_gib():
    """Captures the current process's physical memory footprint (RSS)."""
    return psutil.Process(os.getpid()).memory_info().rss / (1024 ** 3)

def extract_python_code(model_output: str) -> str:
    """Extracts executable python code blocks from markdown responses."""
    code_block = re.search(r"```(?:python)?\s*(.*?)\s*```", model_output, re.DOTALL)
    if code_block:
        return code_block.group(1).strip()
    # Fallback: remove leading natural language if any
    lines = model_output.split("\n")
    code_lines = [l for l in lines if not l.startswith("#") and not l.strip().startswith("Here")]
    return "\n".join(code_lines).strip()

def sandbox_execute_code(code: str, assertions: list, timeout_sec: int = 5) -> str:
    """
    Executes generated code against MBPP unit test assertions in an isolated
    subprocess guarded by a strict timeout to prevent infinite loops.
    """
    assert_str = "\n".join(assertions)
    full_script = f"{code}\n\n{assert_str}\nprint('PASS_ALL')"
    
    try:
        proc = subprocess.run(
            ["python3", "-c", full_script],
            capture_output=True,
            text=True,
            timeout=timeout_sec
        )
        if proc.returncode == 0 and "PASS_ALL" in proc.stdout:
            return "PASS"
        else:
            return "FAIL"
    except subprocess.TimeoutExpired:
        return "TIMEOUT"
    except Exception:
        return "ERROR"
````

````python
import os
import re
import time
import psutil
import subprocess
import requests
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

OLLAMA_API = "http://localhost:11434/api/generate"

def query_ollama(model_name: str, prompt: str):
    payload = {
        "model": model_name,
        "prompt": prompt,
        "stream": False,
        "options": {
            "temperature": 0.0, # Greedy decoding for scientific reproducibility
            "num_predict": 256 # Hard stop at 256 tokens to prevent runaway generation
        }
    }
    resp = requests.post(OLLAMA_API, json=payload, timeout=240)
    return resp.json()

def get_ollama_rss_gib():
    """Captures the physical memory footprint of the active Ollama engine runner."""
    total_rss = 0.0
    found = False
    for proc in psutil.process_iter(['name', 'memory_info']):
        try:
            pname = (proc.info['name'] or '').lower()
            if 'ollama' in pname or 'llama' in pname:
                mem = proc.info.get('memory_info')
                if mem:
                    total_rss += mem.rss / (1024 ** 3)
                    found = True
        except (psutil.NoSuchProcess, psutil.AccessDenied):
            continue
    return total_rss if found else (psutil.Process(os.getpid()).memory_info().rss / (1024 ** 3))

def extract_python_code(text: str) -> str:
    match = re.search(r"```(?:python)?\s*(.*?)\s*```", text, re.DOTALL)
    if match:
        return match.group(1).strip()
    lines = text.split("\n")
    code_lines = [l for l in lines if not l.startswith("#") and not l.strip().startswith("Here")]
    return "\n".join(code_lines).strip()

def sandbox_execute_code(code: str, assertions: list, timeout_sec: int = 5) -> str:
    assert_str = "\n".join(assertions)
    full_script = f"{code}\n\n{assert_str}\nprint('PASS_ALL')"
    try:
        proc = subprocess.run(
            ["python3", "-c", full_script],
            capture_output=True,
            text=True,
            timeout=timeout_sec
        )
        if proc.returncode == 0 and "PASS_ALL" in proc.stdout:
            return "PASS"
        return "FAIL (Assertion/Syntax)"
    except subprocess.TimeoutExpired:
        return "TIMEOUT (Loop Guard)"
    except Exception as e:
        return f"ERROR ({e.__class__.__name__})"

def evaluate_ollama_mbpp(model_name: str, mbpp_dataset, num_samples: int = 5):
    print(f"\n[*] Evaluating: {model_name} on ThinkPad P14s...")
    
    # Warm-up run to eliminate SSD paging penalty
    print("    [+] Running warm-up query (cold-start mitigation)...", end="", flush=True)
    _ = query_ollama(model_name, "def warm_up(): pass")
    print(" [DONE]")
    
    results = []
    test_samples = mbpp_dataset.select(range(min(num_samples, len(mbpp_dataset))))
    
    for idx, item in enumerate(test_samples):
        task_id = item["task_id"]
        # Handle both 'prompt' (sanitized split) and 'text' (raw split)
        problem_desc = item.get("prompt") or item.get("text")
        
        prompt = (
            f"Write a Python function to solve this problem:\n{problem_desc}\n"
            f"Output ONLY executable Python code enclosed inside a ```python ... ``` block."
        )
        test_assertions = item["test_list"]
        
        # Query Ollama
        raw = query_ollama(model_name, prompt)
        
        # Calculate tokens/sec using Ollama's C++ hardware timers
        eval_count = raw.get("eval_count", 0)
        eval_dur_ns = raw.get("eval_duration", 1)
        tok_sec = eval_count / (eval_dur_ns / 1e9) if eval_dur_ns > 0 else 0
        
        rss_gib = get_ollama_rss_gib()
        
        # Sandbox execution
        code = extract_python_code(raw.get("response", ""))
        status = sandbox_execute_code(code, test_assertions, timeout_sec=5)
        
        results.append({
            "task_id": task_id,
            "tok_sec": tok_sec,
            "rss_gib": rss_gib,
            "status": status
        })
        print(f"    Task {idx+1:02d}/{num_samples} (ID {task_id}): [{status.split()[0]}] -> {tok_sec:.2f} tok/s")
        
    avg_speed = sum(r['tok_sec'] for r in results) / len(results)
    pass_rate = (sum(1 for r in results if r['status'] == "PASS") / len(results)) * 100
    peak_rss = max(r['rss_gib'] for r in results)
    
    return {
        "Model": model_name,
        "Peak RSS (GiB)": round(peak_rss, 2),
        "Throughput (tok/s)": round(avg_speed, 2),
        "MBPP pass@1 (%)": round(pass_rate, 1)
    }
````

````python
# Cell 4: WikiText-2 Perplexity Evaluator

def evaluate_perplexity(model, tokenizer, max_length=512, stride=256, max_eval_tokens=10000):
    """
    Computes cross-entropy loss over the WikiText-2 validation set
    using a sliding window to evaluate grammatical fluency/language degradation.
    """
    print("    [+] Loading WikiText-2 validation set...")
    wikitext = load_dataset("wikitext", "wikitext-2-raw-v1", split="validation")
    text = "\n\n".join(wikitext["text"])
    
    encodings = tokenizer(text, return_tensors="pt")
    seq_len = min(encodings.input_ids.size(1), max_eval_tokens)
    input_ids = encodings.input_ids[:, :seq_len].to(device)
    
    nlls = []
    print(f"    [+] Evaluating perplexity across {seq_len} tokens...")
    for begin_loc in range(0, seq_len, stride):
        end_loc = min(begin_loc + max_length, seq_len)
        trg_len = end_loc - begin_loc
        batch_input_ids = input_ids[:, begin_loc:end_loc]
        target_ids = batch_input_ids.clone()
        target_ids[:, :-trg_len] = -100  # Mask context tokens
        
        with torch.no_grad():
            outputs = model(batch_input_ids, labels=target_ids)
            neg_log_likelihood = outputs.loss
            
        nlls.append(neg_log_likelihood)
        if end_loc == seq_len:
            break
            
    ppl = torch.exp(torch.stack(nlls).mean()).item()
    print(f"    [+] WikiText-2 Perplexity: {ppl:.2f}")
    return ppl
````

````python
from datasets import load_dataset
print("[*] Loading sanitized MBPP dataset from Hugging Face...")
mbpp_dataset = load_dataset("google-research-datasets/mbpp", "sanitized", split="test")
print(f"[✓] Loaded {len(mbpp_dataset)} coding tasks.")
````

````python
# Models to benchmark currently available in your Ollama library
# You can add "llama3.2:3b" or quantized variants once pulled
models_to_test = [
    "llama3.2:3b-text-q8_0",
    "llama3.2:3b"
]

summary_records = []
N_SAMPLES = 5  # Quick 5-task run for your meeting demo

for m in models_to_test:
    metrics = evaluate_ollama_mbpp(m, mbpp_dataset, num_samples=N_SAMPLES)
    summary_records.append(metrics)

df_results = pd.DataFrame(summary_records)
print("\n" + "="*60 + "\nRESULTS SUMMARY\n" + "="*60)
print(df_results.to_markdown(index=False))
````
